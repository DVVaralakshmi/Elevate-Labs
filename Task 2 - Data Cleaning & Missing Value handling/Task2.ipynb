# Task 2: Data Cleaning & Missing Value Handling
## Dataset 1: House Prices Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load dataset
df = pd.read_csv("house_prices.csv")

# Identify missing values
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt

# Visualize missing values using a bar chart
df.isnull().sum().plot(kind='bar')
plt.title("Missing Data Pattern")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.show()
```

```python
import numpy as np

# Apply median imputation safely (future-proof)
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
# Apply mode imputation for categorical columns (safe version)
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Calculate missing value percentage for each column
missing_percentage = df.isnull().mean() * 100

# Drop columns with more than 50% missing values
df = df.loc[:, missing_percentage <= 50]
```

```python
# Validate missing values after cleaning
df.isnull().sum()
```

```python
print("Dataset shape after cleaning:")
df.shape
```

```python
# Final dataset preview
df.head()
```

### Before vs After Comparison

- Dataset size reduced by removing columns with extremely high missing values
- All missing values have been handled successfully
- Overall data quality has improved
- Dataset is now ready for Machine Learning tasks
