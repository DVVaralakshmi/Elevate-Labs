# 📊 Task 4 – Feature Encoding & Scaling

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

df = pd.read_csv('adult_income.csv')
df.head()
```

## Identify categorical and numerical features

```python
categorical_features = df.select_dtypes(include="object").columns
numerical_features = df.select_dtypes(exclude="object").columns

categorical_features, numerical_features
```

## Apply Label Encoding where order exists

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

# Example: target variable
df["income"] = le.fit_transform(df["income"])

df.head()
```

## Apply One-Hot Encoding where order does not exist

```python
df_encoded = pd.get_dummies(df, columns=categorical_features.drop("income"), drop_first=True)

df_encoded.head()
```

## Scale numerical features using StandardScaler

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df_encoded[numerical_features] = scaler.fit_transform(df_encoded[numerical_features])

df_encoded.head()
```

## Compare model readiness before and after scaling

```python
print("Before preprocessing shape:", df.shape)
print("After preprocessing shape:", df_encoded.shape)

df.describe(), df_encoded.describe()
```

## Explain impact of scaling on ML algorithms

### 🔑 Key Points on Feature Scaling

- Scaling improves convergence speed  
- Prevents features with large values from dominating the model  
- Essential for distance-based models like **KNN, SVM, and K-Means**  
- Improves model stability and accuracy  

## Save processed dataset

```python
df_encoded.to_csv("adult_income_processed.csv", index=False)

print("Processed dataset saved successfully.")
```
