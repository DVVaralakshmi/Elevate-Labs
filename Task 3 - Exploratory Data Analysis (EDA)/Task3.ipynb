# 📊 Task 3 – Exploratory Data Analysis (EDA)

This notebook demonstrates EDA using **Iris** and **Netflix** datasets.

## 1️⃣ Plot distribution of numerical features using histograms
**Purpose:** Understand data spread, skewness, and range.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('iris.csv')

df.hist(figsize=(10,6))
plt.suptitle('Distribution of Numerical Features', fontsize=14)
plt.show()
```

🔍 **Insight:**
- Features are mostly normally distributed
- Petal features show clear separability

## 2️⃣ Analyze categorical features using count plots
**Purpose:** Understand category frequency and balance.

```python
import seaborn as sns

netflix = pd.read_csv('netflix_titles.csv')

sns.countplot(data=netflix, x='type')
plt.title('Count of Movies vs TV Shows')
plt.show()
```

🔍 **Insight:**
- Movies dominate Netflix content
- Indicates movie-focused strategy

## 3️⃣ Use box plots to identify outliers
**Purpose:** Detect extreme values.

```python
df.boxplot(figsize=(10,6))
plt.title('Outlier Detection using Box Plot')
plt.show()
```

🔍 **Insight:**
- Sepal width shows mild outliers
- Petal features are consistent

## 4️⃣ Plot correlation heatmap
**Purpose:** Identify relationships between features.

```python
# Select only numerical columns for correlation
numeric_df = df.select_dtypes(include=['number'])

plt.figure(figsize=(8,6))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

🔍 **Insight:**
- Strong correlation between petal length & width
- Weak correlation with sepal features

## 5️⃣ Key Observations
- Histograms show distribution
- Count plots show imbalance
- Box plots show outliers
- Heatmap shows relationships

## 6️⃣ Important Features for Prediction
- Petal Length
- Petal Width

**Reason:** Strong correlation and class separation

## 7️⃣ Final Summary
- Dataset is clean
- Petal features are most predictive
- Minimal outliers
- Strong correlations
